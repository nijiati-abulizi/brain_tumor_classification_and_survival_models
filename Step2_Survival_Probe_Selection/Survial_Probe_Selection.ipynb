````python
# %%
import os
import pickle
import warnings
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, KFold
from lifelines.statistics import logrank_test
from lifelines import CoxPHFitter
from sksurv.linear_model import CoxnetSurvivalAnalysis
from sksurv.metrics import concordance_index_censored, cumulative_dynamic_auc
import optuna

from sklearn.preprocessing import StandardScaler
from collections import Counter

warnings.filterwarnings("ignore")
random_state = 42
results_tracker = {}

# -----------------------------------------------------------------------------
# Data Loading & Preprocessing
# -----------------------------------------------------------------------------
def load_data(base_path):
    file_paths = {
        "clinical_data": os.path.join(base_path, "2019_TCGA-CDR-SupplementalTableS1.xlsx"),
        "new_labels": os.path.join(base_path, "Matrix_WHO2021.csv"),
        "gbm_data": os.path.join(base_path, "GBM_450K_Filtered_X-Y,SNPs,Non-overlap850K.csv"),
        "lgg_data": os.path.join(base_path, "LGG-450K_Filtered_X-Y,SNPs,Non-overlap850K.csv"),
    }
    for key, path in file_paths.items():
        if not os.path.exists(path):
            print(f"❌ Error: Missing {key} file: {path}")
            return None

    print("📂 Loading data...")
    main_data_df = pd.read_excel(file_paths["clinical_data"])
    new_categories = pd.read_csv(file_paths["new_labels"])
    gbm_data = pd.read_csv(file_paths["gbm_data"])
    lgg_data = pd.read_csv(file_paths["lgg_data"])
    print("✅ Data Loaded Successfully!")
    results_tracker["data_load"] = {
        "clinical_data": main_data_df.shape,
        "new_labels": new_categories.shape,
        "gbm_data": gbm_data.shape,
        "lgg_data": lgg_data.shape
    }
    return main_data_df, new_categories, gbm_data, lgg_data

def preprocess_methylation_data(df):
    """
    1) Transpose so that rows become samples, columns become CpG features.
    2) Convert Beta values to M-values: M = log2((Beta + eps)/(1 - Beta + eps)).
    """
    df_trns = df.T
    df_trns.columns = df_trns.iloc[0]   
    df_trns = df_trns[1:]              
    df_trns = df_trns.apply(pd.to_numeric, errors="coerce")
    eps = 1e-5
    df_trns = np.log2((df_trns + eps) / (1 - df_trns + eps))
    return df_trns

def merge_data(main_data_df, new_categories, gbm_data, lgg_data):
    main_data_df.rename(columns={'bcr_patient_barcode': 'Patient_ID'}, inplace=True)
    filtered_data_df = main_data_df[main_data_df['type'].isin(['GBM', 'LGG'])]
    merged_clinical_data = pd.merge(filtered_data_df, new_categories, on='Patient_ID', how='inner')
    gbm_trns_final = preprocess_methylation_data(gbm_data)
    lgg_trns_final = preprocess_methylation_data(lgg_data)
    merged_methylation_data = pd.concat([gbm_trns_final, lgg_trns_final])
    merged_methylation_data.rename(columns={'Index': 'Patient_ID'}, inplace=True)
    final_data = pd.merge(
        merged_methylation_data,
        merged_clinical_data,
        left_index=True,
        right_on='Patient_ID',
        how='inner'
    )
    print(f"✅ Final Merged Data Shape: {final_data.shape}")
    results_tracker["merged_data"] = final_data.shape
    return final_data

# -----------------------------------------------------------------------------
# Prepare Data with Cancer Type as a Categorical Variable & Composite Stratification
# -----------------------------------------------------------------------------
def prepare_data(df):
    # Remove unclassified samples
    df = df[df['classification.2021_simplified.labels'] != 'unclassified']
    
    # Check that exactly 3 unique cancer types remain
    cancer_types = df['classification.2021_simplified.labels'].unique()
    if len(cancer_types) != 3:
        print(f"Warning: Expected 3 cancer types but got {len(cancer_types)}: {cancer_types}")
    else:
        print("✅ Exactly 3 unique cancer types present.")
    
    df = df.dropna(subset=["OS.time", "OS"])
    df["OS"] = pd.to_numeric(df["OS"], errors="coerce").astype(int)
    
    # Identify methylation features
    methylation_features = [col for col in df.columns if col.startswith("cg")]
    if len(methylation_features) == 0:
        print("❌ No valid methylation features found. Skipping.")
        return None
    
    X_numeric = df[methylation_features].copy()
    # One-hot encode the cancer type; do not drop first so that all 3 types are represented
    cancer_dummies = pd.get_dummies(df['classification.2021_simplified.labels'], prefix='cancer')
    # Combine methylation features with cancer type dummies
    X = pd.concat([X_numeric, cancer_dummies], axis=1)
    
    y = df[['OS.time', 'OS']]
    
    # Create composite stratification variable: combine OS (event) and cancer type
    stratify_var = df["OS"].astype(str) + "_" + df["classification.2021_simplified.labels"]
    
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=random_state, stratify=stratify_var
    )
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
    X_test = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)
    
    print(f"✅ Train: {X_train.shape[0]}, Test: {X_test.shape[0]}")
    return {
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test
    }

def convert_survival_data(y_df):
    return np.array(
        [(bool(event), time) for event, time in zip(y_df["OS"], y_df["OS.time"])],
        dtype=[('event', '?'), ('time', '<f8')]
    )

# -----------------------------------------------------------------------------
# Training & Evaluation Functions (Elastic Net Version)
# -----------------------------------------------------------------------------
def train_lasso_cox_model(X_train, y_train_struct, l1_ratio=0.9, alpha_min_ratio=0.5, n_alphas=50):
    model = CoxnetSurvivalAnalysis(
        l1_ratio=l1_ratio, alpha_min_ratio=alpha_min_ratio, n_alphas=n_alphas
    )
    model.fit(X_train, y_train_struct)
    return model

def extract_significant_features(model, X_train):
    best_alpha_index = np.argmax(np.sum(model.coef_ != 0, axis=0))
    selected_features = X_train.columns[model.coef_[:, best_alpha_index] != 0]
    return selected_features

def evaluate_model(model, X_train, y_train_struct, X_test, y_test_struct):
    train_c_index = model.score(X_train, y_train_struct)
    test_c_index = model.score(X_test, y_test_struct)
    
    times = np.percentile(y_test_struct["time"], np.linspace(5, 95, 10))
    auc, mean_auc = cumulative_dynamic_auc(y_train_struct, y_test_struct, model.predict(X_test), times)
    
    print(f"📈 Harrell’s C-index: Train={train_c_index:.3f}, Test={test_c_index:.3f}")
    print(f"📈 Mean Time-dependent AUC (Test): {mean_auc:.3f}")
    return train_c_index, test_c_index, mean_auc

# ---- Optuna-based CV Tuning (Elastic Net) ----
def cross_val_c_index(X, y_struct, l1_ratio, alpha_min_ratio, n_alphas, n_splits=5):
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    c_indexes = []
    for train_idx, val_idx in kf.split(X):
        X_train_fold = X.iloc[train_idx]
        X_val_fold = X.iloc[val_idx]
        y_train_fold = y_struct[train_idx]
        y_val_fold = y_struct[val_idx]
        
        scaler = StandardScaler()
        X_train_fold_scaled = scaler.fit_transform(X_train_fold)
        X_val_fold_scaled = scaler.transform(X_val_fold)
        
        model = CoxnetSurvivalAnalysis(
            l1_ratio=l1_ratio, alpha_min_ratio=alpha_min_ratio, n_alphas=n_alphas
        )
        model.fit(X_train_fold_scaled, y_train_fold)
        c_indexes.append(model.score(X_val_fold_scaled, y_val_fold))
    return np.mean(c_indexes)

def tune_hyperparameters_optuna(X, y_struct, fixed_l1_ratio, n_trials=20):
    def objective(trial):
        alpha_min_ratio = trial.suggest_float("alpha_min_ratio", 0.001, 1)
        n_alphas = trial.suggest_int("n_alphas", 50, 300)
        return cross_val_c_index(X, y_struct, fixed_l1_ratio, alpha_min_ratio, n_alphas, n_splits=5)
    study = optuna.create_study(direction="maximize")
    study.optimize(objective, n_trials=n_trials)
    return study.best_params

def compute_shap_importance(model, X_subset):
    print(f"🔍 Computing SHAP on {X_subset.shape[1]} features...")
    explainer = shap.Explainer(model.predict, X_subset)
    shap_values = explainer(X_subset)
    shap.summary_plot(shap_values, X_subset, show=False)
    plt.title("SHAP Summary Plot")
    plt.show()

# -----------------------------------------------------------------------------
# Main Workflow (Entire Data with Cancer Type as Predictor)
# -----------------------------------------------------------------------------
def main():
    base_path = "/Users/nijat/Downloads/OneDrive_1_2024-09-21/Data"
    data_loaded = load_data(base_path)
    if data_loaded is None:
        return
    main_data_df, new_categories, gbm_data, lgg_data = data_loaded
    merged_data = merge_data(main_data_df, new_categories, gbm_data, lgg_data)
    
    prepared_data = prepare_data(merged_data)
    if prepared_data is None:
        return
    
    prepared_data["y_train_struct"] = convert_survival_data(prepared_data["y_train"])
    prepared_data["y_test_struct"] = convert_survival_data(prepared_data["y_test"])
    
    X_train = prepared_data["X_train"]
    X_test = prepared_data["X_test"]
    y_train_struct = prepared_data["y_train_struct"]
    y_test_struct = prepared_data["y_test_struct"]
    
    # Loop over candidate l1_ratio values
    candidate_l1_ratios = [1, 0.9, 0.8, 0.7, 0.6, 0.5]
    results = {}
    
    for l1_ratio in candidate_l1_ratios:
        print("\n==========================================")
        print(f"Evaluating for l1_ratio = {l1_ratio}")
        
        # Hyperparameter tuning with fixed l1_ratio and 20 trials
        best_params = tune_hyperparameters_optuna(X_train, y_train_struct, fixed_l1_ratio=l1_ratio, n_trials=20)
        print(f"   Best params for l1_ratio {l1_ratio} -> {best_params}")
        
        # Retrain final model with tuned hyperparameters
        final_model = train_lasso_cox_model(
            X_train, y_train_struct,
            l1_ratio=l1_ratio,
            alpha_min_ratio=best_params["alpha_min_ratio"],
            n_alphas=best_params["n_alphas"]
        )
        train_c, test_c, auc_val = evaluate_model(final_model, X_train, y_train_struct, X_test, y_test_struct)
        print(f"   Final Model for l1_ratio {l1_ratio} -> Train C-index={train_c:.3f}, Test C-index={test_c:.3f}, AUC={auc_val:.3f}")
        
        # Extract significant features (selected probes)
        selected_features = extract_significant_features(final_model, X_train)
        print(f"   -> {len(selected_features)} features retained for l1_ratio {l1_ratio}")
        print("   Selected probes:")
        print(list(selected_features))
        
        # Optionally retrain a subset model on these selected features and compute SHAP
        if len(selected_features) > 0:
            final_model_subset = train_lasso_cox_model(
                X_train[selected_features],
                y_train_struct,
                l1_ratio=l1_ratio,
                alpha_min_ratio=best_params["alpha_min_ratio"],
                n_alphas=best_params["n_alphas"]
            )
            subset_train_c, subset_test_c, subset_auc = evaluate_model(
                final_model_subset,
                X_train[selected_features],
                y_train_struct,
                X_test[selected_features],
                y_test_struct
            )
            print(f"   Subset Model for l1_ratio {l1_ratio} -> Train C-index={subset_train_c:.3f}, Test C-index={subset_test_c:.3f}, AUC={subset_auc:.3f}")
            compute_shap_importance(final_model_subset, X_train[selected_features])
        
        # Store results for this candidate l1_ratio
        results[l1_ratio] = {
            "selected_probes": list(selected_features),
            "num_selected": len(selected_features),
            "train_c_index": train_c,
            "test_c_index": test_c,
            "auc": auc_val,
            "best_params": best_params
        }
    
    # Print summary for each candidate l1_ratio
    print("\n=== Summary of Selected Probes and Metrics for Each l1_ratio ===")
    for l1_ratio, res in results.items():
        print(f"\nl1_ratio = {l1_ratio}:")
        print(f"   Number of selected probes: {res['num_selected']}")
        print(f"   Train C-index: {res['train_c_index']:.3f}")
        print(f"   Test C-index: {res['test_c_index']:.3f}")
        print(f"   AUC: {res['auc']:.3f}")
        print(f"   Best tuning parameters: {res['best_params']}")
        print("   Selected probes:")
        print(res["selected_probes"])
    
    print("\n✅ Workflow Complete!")

if __name__ == "__main__":
    main()
````

````python
# %%
import os
import pickle
import warnings
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt

from itertools import product
from sklearn.model_selection import train_test_split, KFold
from lifelines.statistics import logrank_test
from lifelines import CoxPHFitter
from sksurv.linear_model import CoxnetSurvivalAnalysis
from sksurv.metrics import concordance_index_censored, cumulative_dynamic_auc
import optuna

from sklearn.preprocessing import StandardScaler
from collections import Counter

warnings.filterwarnings("ignore")
random_state = 42

results_tracker = {}

# -----------------------------------------------------------------------------
# Data Loading & Preprocessing
# -----------------------------------------------------------------------------
def load_data(base_path):
    file_paths = {
        "clinical_data": os.path.join(base_path, "2019_TCGA-CDR-SupplementalTableS1.xlsx"),
        "new_labels": os.path.join(base_path, "Matrix_WHO2021.csv"),
        "gbm_data": os.path.join(base_path, "GBM_450K_Filtered_X-Y,SNPs,Non-overlap850K.csv"),
        "lgg_data": os.path.join(base_path, "LGG-450K_Filtered_X-Y,SNPs,Non-overlap850K.csv"),
    }
    for key, path in file_paths.items():
        if not os.path.exists(path):
            print(f"❌ Error: Missing {key} file: {path}")
            return None

    print("📂 Loading data...")
    main_data_df = pd.read_excel(file_paths["clinical_data"])
    new_categories = pd.read_csv(file_paths["new_labels"])
    gbm_data = pd.read_csv(file_paths["gbm_data"])
    lgg_data = pd.read_csv(file_paths["lgg_data"])
    print("✅ Data Loaded Successfully!")

    results_tracker["data_load"] = {
        "clinical_data": main_data_df.shape,
        "new_labels": new_categories.shape,
        "gbm_data": gbm_data.shape,
        "lgg_data": lgg_data.shape
    }
    return main_data_df, new_categories, gbm_data, lgg_data

def preprocess_methylation_data(df):
    """
    1) Transpose so that rows become samples, columns become CpG features.
    2) Convert Beta values to M-values: M = log2((Beta + eps)/(1 - Beta + eps)).
    """

    df_trns = df.T
    df_trns.columns = df_trns.iloc[0]
    df_trns = df_trns[1:]
    df_trns = df_trns.apply(pd.to_numeric, errors="coerce")

    # Convert Beta -> M-values
    eps = 1e-5
    df_trns = np.log2((df_trns + eps) / (1 - df_trns + eps))

    return df_trns

def merge_data(main_data_df, new_categories, gbm_data, lgg_data):
    main_data_df.rename(columns={'bcr_patient_barcode': 'Patient_ID'}, inplace=True)
    filtered_data_df = main_data_df[main_data_df['type'].isin(['GBM', 'LGG'])]
    merged_clinical_data = pd.merge(filtered_data_df, new_categories, on='Patient_ID', how='inner')

    gbm_trns_final = preprocess_methylation_data(gbm_data)
    lgg_trns_final = preprocess_methylation_data(lgg_data)
    merged_methylation_data = pd.concat([gbm_trns_final, lgg_trns_final])
    merged_methylation_data.rename(columns={'Index': 'Patient_ID'}, inplace=True)

    final_data = pd.merge(
        merged_methylation_data,
        merged_clinical_data,
        left_index=True,
        right_on='Patient_ID',
        how='inner'
    )
    print(f"✅ Final Merged Data Shape: {final_data.shape}")
    results_tracker["merged_data"] = final_data.shape
    return final_data

def split_data(df):
    # Filter out "unclassified" and split by cancer subtype
    df = df[df['classification.2021_simplified.labels'] != 'unclassified']
    subsets = {
        'astrocytoma': df[df['classification.2021_simplified.labels'] == 'astrocytoma'],
        'glioblastoma': df[df['classification.2021_simplified.labels'] == 'glioblastoma'],
        'oligodendroglioma': df[df['classification.2021_simplified.labels'] == 'oligodendroglioma']
    }
    print(f"✅ Data Split into: {', '.join(subsets.keys())}")
    results_tracker["split_data"] = {k: v.shape for k, v in subsets.items()}
    return subsets

# -----------------------------------------------------------------------------
# Stratified Train-Test Splitting (70/30) using Censoring Status
# -----------------------------------------------------------------------------
def prepare_subtype_data(df):
    df = df.dropna(subset=["OS.time", "OS"])
    df = df.apply(pd.to_numeric, errors="coerce")
    df["OS"] = df["OS"].astype(int)

    # Identify methylation features
    methylation_features = [col for col in df.columns if col.startswith("cg")]
    if len(methylation_features) == 0:
        print("❌ No valid methylation features found. Skipping.")
        return None

    X = df[methylation_features]
    y = df[['OS.time', 'OS']]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=random_state, stratify=y["OS"]
    )

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
    X_test  = pd.DataFrame(X_test_scaled,  columns=X_test.columns,  index=X_test.index)

    print(f"✅ Train: {X_train.shape[0]}, Test: {X_test.shape[0]}")
    return {
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test
    }

def convert_survival_data(y_df):
    return np.array(
        [(bool(event), time) for event, time in zip(y_df["OS"], y_df["OS.time"])],
        dtype=[('event', '?'), ('time', '<f8')]
    )

# -----------------------------------------------------------------------------
# Training & Evaluation Functions
# -----------------------------------------------------------------------------
def train_lasso_cox_model(X_train, y_train_struct, alpha_min_ratio=0.5, n_alphas=50):
    model = CoxnetSurvivalAnalysis(
        l1_ratio=1, alpha_min_ratio=alpha_min_ratio, n_alphas=n_alphas
    )
    model.fit(X_train, y_train_struct)
    return model

def extract_significant_features(coxnet, X_train):
    best_alpha_index = np.argmax(np.sum(coxnet.coef_ != 0, axis=0))
    selected_features = X_train.columns[coxnet.coef_[:, best_alpha_index] != 0]
    return selected_features

def evaluate_model(coxnet, X_train, y_train_struct, X_test, y_test_struct):
    train_c_index = coxnet.score(X_train, y_train_struct)
    test_c_index  = coxnet.score(X_test,  y_test_struct)

    # Time-dependent AUC on test set
    times = np.percentile(y_test_struct["time"], np.linspace(5, 95, 10))
    auc, mean_auc = cumulative_dynamic_auc(
        y_train_struct, y_test_struct, coxnet.predict(X_test), times
    )

    print(f"📈 Harrell’s C-index: Train={train_c_index:.3f}, Test={test_c_index:.3f}")
    print(f"📈 Mean Time-dependent AUC (Test): {mean_auc:.3f}")
    return train_c_index, test_c_index, mean_auc

# ---- Optuna-based CV Tuning ----
def cross_val_c_index(X, y_struct, alpha_min_ratio, n_alphas, n_splits=5):
    # l1_ratio=1
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    c_indexes = []

    for train_idx, val_idx in kf.split(X):
        X_train_fold = X.iloc[train_idx]
        X_val_fold   = X.iloc[val_idx]
        y_train_fold = y_struct[train_idx]
        y_val_fold   = y_struct[val_idx]

        scaler = StandardScaler()
        X_train_fold_scaled = scaler.fit_transform(X_train_fold)
        X_val_fold_scaled   = scaler.transform(X_val_fold)

        model = CoxnetSurvivalAnalysis(
            l1_ratio=1, alpha_min_ratio=alpha_min_ratio, n_alphas=n_alphas
        )
        model.fit(X_train_fold_scaled, y_train_fold)
        c_indexes.append(model.score(X_val_fold_scaled, y_val_fold))

    return np.mean(c_indexes)

def tune_hyperparameters_optuna(X, y_struct, n_trials=20):
    def objective(trial):
        alpha_min_ratio = trial.suggest_float("alpha_min_ratio", 0.001, 1)
        n_alphas        = trial.suggest_int("n_alphas", 50, 300)
        return cross_val_c_index(X, y_struct, alpha_min_ratio, n_alphas, n_splits=5)

    study = optuna.create_study(direction="maximize")
    study.optimize(objective, n_trials=n_trials)
    return study.best_params

def compute_shap_importance(model, X_subset):
    print(f"🔍 Computing SHAP on {X_subset.shape[1]} features...")
    explainer = shap.Explainer(model.predict, X_subset)
    shap_values = explainer(X_subset)
    shap.summary_plot(shap_values, X_subset, show=False)
    plt.title("SHAP Summary Plot")
    plt.show()

# -----------------------------------------------------------------------------
# Main Workflow
# -----------------------------------------------------------------------------
def main():
    base_path = "/Users/nijat/Downloads/OneDrive_1_2024-09-21/Data"
    data_loaded = load_data(base_path)
    if data_loaded is None:
        return

    main_data_df, new_categories, gbm_data, lgg_data = data_loaded
    merged_data = merge_data(main_data_df, new_categories, gbm_data, lgg_data)
    subsets = split_data(merged_data)

    split_data_dict = {}
    for label, df in subsets.items():
        print(f"\n🔹 Processing subtype={label}")
        subtype_data = prepare_subtype_data(df)
        if subtype_data is None:
            continue
        # Convert survival data
        subtype_data["y_train_struct"] = convert_survival_data(subtype_data["y_train"])
        subtype_data["y_test_struct"]  = convert_survival_data(subtype_data["y_test"])
        split_data_dict[label] = subtype_data

    # Dictionary to store selected probes for each cancer type
    selected_probes_by_cancer = {}

    # Train/Eval per subtype
    for label, data in split_data_dict.items():
        print(f"\n==============================")
        print(f"🔹 Subtype: {label}")

        X_train = data["X_train"]
        X_test  = data["X_test"]
        y_train_struct = data["y_train_struct"]
        y_test_struct  = data["y_test_struct"]

        # Baseline
        print("🔸 Baseline: alpha_min_ratio=0.5, n_alphas=50")
        baseline_model = train_lasso_cox_model(
            X_train, y_train_struct, alpha_min_ratio=0.5, n_alphas=50
        )
        train_c, test_c, auc_val = evaluate_model(
            baseline_model, X_train, y_train_struct, X_test, y_test_struct
        )
        print(f"   Baseline -> Train C-index={train_c:.3f}, Test C-index={test_c:.3f}, AUC={auc_val:.3f}")

        # Hyperparameter Tuning with Optuna
        print("🔸 Tuning with Optuna + 5-fold CV...")
        best_params = tune_hyperparameters_optuna(X_train, y_train_struct, n_trials=20)
        print(f"   Best params -> {best_params}")

        # Final Model
        final_model = train_lasso_cox_model(
            X_train, y_train_struct,
            alpha_min_ratio=best_params["alpha_min_ratio"],
            n_alphas=best_params["n_alphas"]
        )
        final_train_c, final_test_c, final_auc = evaluate_model(
            final_model, X_train, y_train_struct, X_test, y_test_struct
        )
        print(f"   Final Model -> Train C-index={final_train_c:.3f}, Test C-index={final_test_c:.3f}, AUC={final_auc:.3f}")

        # Extract significant features
        selected_features = extract_significant_features(final_model, X_train)
        print(f"   -> {len(selected_features)} features retained")
        selected_probes_by_cancer[label] = list(selected_features)
        print(f"   Selected probes for {label}:")
        print(list(selected_features))

        # Subset Model & SHAP
        if len(selected_features) > 0:
            final_model_subset = train_lasso_cox_model(
                X_train[selected_features],
                y_train_struct,
                alpha_min_ratio=best_params["alpha_min_ratio"],
                n_alphas=best_params["n_alphas"]
            )
            subset_train_c, subset_test_c, subset_auc = evaluate_model(
                final_model_subset,
                X_train[selected_features],
                y_train_struct,
                X_test[selected_features],
                y_test_struct
            )
            print(f"   Subset Model -> Train C-index={subset_train_c:.3f}, Test C-index={subset_test_c:.3f}, AUC={subset_auc:.3f}")
            compute_shap_importance(final_model_subset, X_train[selected_features])

    # Print final summary of selected probes
    print("\n=== Selected Probes Summary ===")
    for label, probes in selected_probes_by_cancer.items():
        print(f"Subtype: {label} - {len(probes)} probes selected.")
        print(probes)
        print()

    # Compute union of all selected probes
    all_probes = []
    for probes in selected_probes_by_cancer.values():
        all_probes.extend(probes)
    unique_probes = set(all_probes)
    probe_counts = Counter(all_probes)

    print("Union of all selected probes:")
    print(f"Total unique selected probes (union): {len(unique_probes)}")

    print("\n✅ Workflow Complete!")

if __name__ == "__main__":
    main()
````
