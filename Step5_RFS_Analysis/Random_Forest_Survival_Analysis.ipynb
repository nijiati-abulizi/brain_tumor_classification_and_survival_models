```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sksurv.ensemble import RandomSurvivalForest
from sksurv.metrics import cumulative_dynamic_auc, integrated_brier_score
from sklearn.model_selection import train_test_split, KFold
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test
import optuna
import time

# -------------------------------
# 1. Load the Merged Survival Data
# -------------------------------
data_path = "/Users/nijat/Desktop/Final/merged_survival_analysis.csv"
df = pd.read_csv(data_path)
print("Merged survival analysis data shape:", df.shape)

# -------------------------------
# 2. Define Predictor Groups and Outcome Columns
# -------------------------------
# (a) Methylation features: all columns starting with "cg"
methylation_cols = [col for col in df.columns if col.startswith("cg")]

# (b) Gene data columns (provided list)
gene_cols = ["MCL1", "MYCN", "GLI2", "EGFR", "CDK6", "MET", "FGFR1", "MYC", 
             "CDKN2A.B", "PTCH1", "PTEN", "CCND1", "KRAS", "CDK4", "MDM2", 
             "RB1", "TP53", "NF1", "ERBB2", "CCNE1"]

# (c) MGMT column (assumed to be present)
mgmt_col = ["MGMT"]

# (d) Combined feature set
combined_cols = methylation_cols + gene_cols + mgmt_col

# (e) Survival outcome columns and cancer type for stratification.
surv_cols = ["OS", "OS.time", "classification.2021_simplified.labels"]

# Drop rows with missing values for all predictors and survival columns.
df = df.dropna(subset=combined_cols + surv_cols)
print("Data shape after dropping missing values:", df.shape)

# Create a structured array for the survival outcome.
y_all = np.array([(bool(row["OS"]), row["OS.time"]) for idx, row in df.iterrows()],
                 dtype=[("event", "bool"), ("time", "float")])
# Cancer type for stratification.
cancer_type_all = df["classification.2021_simplified.labels"].astype(str)

# Define a dictionary for each predictor group.
predictor_groups = {
    "Methylation": methylation_cols,
    "Gene": gene_cols,
    "MGMT": mgmt_col,
    "Combined": combined_cols
}

# Dictionary to store performance metrics.
results_dict = {}

# -------------------------------
# 3. Loop over Each Predictor Group
# -------------------------------
for group_name, predictors in predictor_groups.items():
    print(f"\n--- Processing Predictor Group: {group_name} ---")
    
    # Create feature matrix for the current group.
    X_all = df[predictors].copy()
    
    # Use the same survival outcome and cancer type.
    y = y_all
    cancer_type = cancer_type_all

    # -------------------------------
    # 4. Train-Test Split (Stratified)
    # -------------------------------
    stratify_label = cancer_type + "_" + df["OS"].astype(str)
    X_train, X_test, y_train, y_test, cancer_train, cancer_test = train_test_split(
        X_all, y, cancer_type, test_size=0.3, random_state=42, stratify=stratify_label)
    print("Training set shape:", X_train.shape)
    print("Test set shape:", X_test.shape)
    
    # -------------------------------
    # 5. Determine Time Limits for Evaluation
    # -------------------------------
    test_event_times = np.array([row["time"] for row in y_test if row["event"]])
    if len(test_event_times) == 0:
        print(f"No events found in test set for {group_name}; skipping...")
        continue
    max_event_time = test_event_times.max()
    time_points = np.array([365, 1095, 1825], dtype=float)  # 1, 3, and 5 years in days
    if time_points[-1] > max_event_time:
        print("Warning: 5-year time point exceeds maximum event time; adjusting it.")
        time_points[-1] = max_event_time * 0.9
    print("Evaluation time points (days):", time_points)
    
    # -------------------------------
    # 6. Hyperparameter Tuning with Optuna
    # -------------------------------
    def objective(trial):
        n_estimators = trial.suggest_int("n_estimators", 50, 200)
        min_samples_split = trial.suggest_int("min_samples_split", 2, 20)
        min_samples_leaf = trial.suggest_int("min_samples_leaf", 1, 20)
        max_features = trial.suggest_categorical("max_features", ["sqrt", "log2", None])
        
        rsf = RandomSurvivalForest(
            n_estimators=n_estimators,
            min_samples_split=min_samples_split,
            min_samples_leaf=min_samples_leaf,
            max_features=max_features,
            n_jobs=-1,
            random_state=42
        )
        
        cv = KFold(n_splits=3, shuffle=True, random_state=42)
        c_indexes = []
        for train_idx, val_idx in cv.split(X_train):
            X_tr = X_train.iloc[train_idx]
            y_tr = y_train[train_idx]
            X_val = X_train.iloc[val_idx]
            y_val = y_train[val_idx]
            rsf.fit(X_tr, y_tr)
            c_indexes.append(rsf.score(X_val, y_val))
        return np.mean(c_indexes)
    
    print("Starting hyperparameter tuning with Optuna...")
    study = optuna.create_study(direction="maximize")
    study.optimize(objective, n_trials=50, show_progress_bar=True)
    best_params = study.best_trial.params
    print("Best hyperparameters:", best_params)
    
    # -------------------------------
    # 7. Train Final RSF Model with Best Parameters
    # -------------------------------
    rsf_final = RandomSurvivalForest(
        n_estimators=best_params["n_estimators"],
        min_samples_split=best_params["min_samples_split"],
        min_samples_leaf=best_params["min_samples_leaf"],
        max_features=best_params["max_features"],
        n_jobs=-1,
        random_state=42
    )
    start_model_time = time.time()
    rsf_final.fit(X_train, y_train)
    training_time = time.time() - start_model_time
    print("Final RSF model trained in {:.2f} seconds.".format(training_time))
    
    # -------------------------------
    # 8. Model Evaluation
    # -------------------------------
    # (a) C-index
    c_index = rsf_final.score(X_test, y_test)
    print("C-index on test set:", c_index)
    
    # (b) Time-dependent AUC
    risk_scores_test = rsf_final.predict(X_test)
    auc, mean_auc = cumulative_dynamic_auc(y_train, y_test, risk_scores_test, time_points)
    for t, a in zip(time_points, auc):
        print(f"AUC at {t:.0f} days (~{t/365:.1f} years): {a}")
    
    # (c) Integrated Brier Score (IBS)
    surv_funcs = rsf_final.predict_survival_function(X_test)
    pred_surv = np.row_stack([fn(time_points) for fn in surv_funcs])
    ibs = integrated_brier_score(y_train, y_test, pred_surv, time_points)
    print("Integrated Brier Score (IBS):", ibs)
    
    # Save metrics for later comparison.
    results_dict[group_name] = {
        "c_index": c_index,
        "mean_auc": mean_auc,
        "ibs": ibs,
        "training_time": training_time,
        "best_params": best_params
    }

# -------------------------------
# 9. Compare Performance Across Predictor Groups
# -------------------------------
print("\n=== Performance Comparison ===")
for group, metrics in results_dict.items():
    print(f"\nGroup: {group}")
    print(f"  C-index: {metrics['c_index']:.4f}")
    print(f"  Mean AUC: {metrics['mean_auc']:.4f}")
    print(f"  Integrated Brier Score (IBS): {metrics['ibs']:.4f}")
    print(f"  Training time (s): {metrics['training_time']:.2f}")
    print(f"  Best parameters: {metrics['best_params']}")

# -------------------------------
# 10. Plot Comparison of Metrics
# -------------------------------
groups = list(results_dict.keys())
c_indexes = [results_dict[group]["c_index"] for group in groups]
mean_aucs = [results_dict[group]["mean_auc"] for group in groups]
ibs_values = [results_dict[group]["ibs"] for group in groups]

# Create a DataFrame for visualization
metrics_df = pd.DataFrame({
    "Group": groups,
    "C-index": c_indexes,
    "Mean AUC": mean_aucs,
    "IBS": ibs_values
})
print(metrics_df)

# Plotting the metrics side by side.
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

# C-index Comparison
axs[0].bar(groups, c_indexes, color="skyblue")
axs[0].set_title("C-index Comparison")
axs[0].set_ylabel("C-index")
axs[0].set_ylim(0, 1)

# Mean AUC Comparison
axs[1].bar(groups, mean_aucs, color="lightgreen")
axs[1].set_title("Mean AUC Comparison")
axs[1].set_ylabel("Mean AUC")
axs[1].set_ylim(0, 1)

# Integrated Brier Score (IBS) Comparison
axs[2].bar(groups, ibs_values, color="salmon")
axs[2].set_title("Integrated Brier Score (IBS) Comparison")
axs[2].set_ylabel("IBS")
axs[2].set_ylim(0, max(ibs_values)*1.1)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sksurv.ensemble import RandomSurvivalForest
from sksurv.metrics import cumulative_dynamic_auc
from sklearn.model_selection import train_test_split, KFold
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test
import optuna
import time

# -------------------------------
# 1. Load the Merged Survival Data
# -------------------------------
data_path = "/Users/nijat/Desktop/Final/merged_survival_analysis.csv"
df = pd.read_csv(data_path)
print("Merged survival analysis data shape:", df.shape)

# -------------------------------
# 2. Define Features and Outcome
# -------------------------------

# (a) Methylation features: all columns starting with "cg"
methylation_cols = [col for col in df.columns if col.startswith("cg")]

# (b) Gene data columns
gene_cols = ["MCL1", "MYCN", "GLI2", "EGFR", "CDK6", "MET", "FGFR1", "MYC", 
             "CDKN2A.B", "PTCH1", "PTEN", "CCND1", "KRAS", "CDK4", "MDM2", 
             "RB1", "TP53", "NF1", "ERBB2", "CCNE1"]

# (c) MGMT column 
feature_cols = methylation_cols + gene_cols + ["MGMT"]

# (d) Survival outcome columns: OS (event indicator) and OS.time (time in days)
# Also keep cancer type for stratification.
surv_cols = ["OS", "OS.time", "classification.2021_simplified.labels"]

df = df.dropna(subset=feature_cols + surv_cols)
print("Data shape after dropping missing values:", df.shape)

# Create a structured array for the survival outcome required by scikit-survival.
y = np.array([(bool(row["OS"]), row["OS.time"]) for idx, row in df.iterrows()],
             dtype=[("event", "bool"), ("time", "float")])
X = df[feature_cols].copy()
cancer_type = df["classification.2021_simplified.labels"].astype(str)

# -------------------------------
# 3. Train-Test Split (Stratified)
# -------------------------------
stratify_label = cancer_type + "_" + df["OS"].astype(str)
X_train, X_test, y_train, y_test, cancer_train, cancer_test = train_test_split(
    X, y, cancer_type, test_size=0.3, random_state=42, stratify=stratify_label)
print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)

# -------------------------------
# 4. Determine Time Limits for Evaluation
# -------------------------------
test_event_times = np.array([row["time"] for row in y_test if row["event"]])
if len(test_event_times) == 0:
    raise ValueError("No events found in test set!")
max_event_time = test_event_times.max()
print("Maximum event time among events in test set:", max_event_time, "days")

time_points = np.array([365, 1095, 1825], dtype=float)  
if time_points[-1] > max_event_time:
    print("Warning: 5-year time point (1825 days) exceeds maximum event time; adjusting it.")
    time_points[-1] = max_event_time * 0.9
print("Evaluation time points (days):", time_points)

# -------------------------------
# 5. Hyperparameter Tuning with Optuna
# -------------------------------
def objective(trial):
    n_estimators = trial.suggest_int("n_estimators", 50, 200)
    min_samples_split = trial.suggest_int("min_samples_split", 2, 20)
    min_samples_leaf = trial.suggest_int("min_samples_leaf", 1, 20)
    max_features = trial.suggest_categorical("max_features", ["sqrt", "log2", None])
    
    rsf = RandomSurvivalForest(
        n_estimators=n_estimators,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        max_features=max_features,
        n_jobs=-1,
        random_state=42
    )
    
    cv = KFold(n_splits=3, shuffle=True, random_state=42)
    c_indexes = []
    for train_idx, val_idx in cv.split(X_train):
        X_tr = X_train.iloc[train_idx]
        y_tr = y_train[train_idx]
        X_val = X_train.iloc[val_idx]
        y_val = y_train[val_idx]
        rsf.fit(X_tr, y_tr)
        c_index = rsf.score(X_val, y_val)
        c_indexes.append(c_index)
    return np.mean(c_indexes)

print("Starting hyperparameter tuning with Optuna...")
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=100, show_progress_bar=True)
print("Best hyperparameters:", study.best_trial.params)

# -------------------------------
# 6. Train Final RSF Model with Best Parameters
# -------------------------------
best_params = study.best_trial.params
rsf_final = RandomSurvivalForest(
    n_estimators=best_params["n_estimators"],
    min_samples_split=best_params["min_samples_split"],
    min_samples_leaf=best_params["min_samples_leaf"],
    max_features=best_params["max_features"],
    n_jobs=-1,
    random_state=42
)
start_model_time = time.time()
rsf_final.fit(X_train, y_train)
print("Final Random Survival Forest model trained in {:.2f} seconds.".format(time.time()-start_model_time))

# -------------------------------
# 7. Model Evaluation
# -------------------------------
c_index_final = rsf_final.score(X_test, y_test)
print("Overall C-index on test set:", c_index_final)
risk_scores_test = rsf_final.predict(X_test)
auc, mean_auc = cumulative_dynamic_auc(y_train, y_test, risk_scores_test, time_points)
for t, a in zip(time_points, auc):
    print(f"AUC at {t:.0f} days (~{t/365:.1f} years): {a}")

# -------------------------------
# 8. Combined Risk Stratification and Kaplan-Meier Curves with Data-Driven Cutpoint
# -------------------------------
def find_optimal_cutpoint(df, risk_column="risk", time_column="OS.time", event_column="OS"):
    """
    Tests candidate thresholds (from the 10th to 90th percentile) and returns the threshold 
    that minimizes the log-rank test p-value.
    """
    candidate_thresholds = np.percentile(df[risk_column], np.arange(10, 91, 5))
    best_p = 1.0
    best_threshold = None
    for threshold in candidate_thresholds:
        df_temp = df.copy()
        df_temp["risk_group"] = df_temp[risk_column].apply(lambda x: "high" if x >= threshold else "low")
        # Ensure both groups have events; skip if one group has none.
        if df_temp[df_temp["risk_group"]=="low"][event_column].sum() == 0 or \
           df_temp[df_temp["risk_group"]=="high"][event_column].sum() == 0:
            continue
        results = logrank_test(
            df_temp[df_temp["risk_group"]=="low"][time_column],
            df_temp[df_temp["risk_group"]=="high"][time_column],
            event_observed_A=df_temp[df_temp["risk_group"]=="low"][event_column],
            event_observed_B=df_temp[df_temp["risk_group"]=="high"][event_column]
        )
        if results.p_value < best_p:
            best_p = results.p_value
            best_threshold = threshold
    return best_threshold, best_p

# Build test_results DataFrame with risk scores, OS.time, OS, and cancer type.
test_results = X_test.copy()
test_results["risk"] = risk_scores_test
test_results["OS.time"] = [row["time"] for row in y_test]
test_results["OS"] = [row["event"] for row in y_test]
test_results["cancer_type"] = cancer_test.values

plt.figure(figsize=(10, 6))
unique_cancers = test_results["cancer_type"].unique()
for ct in unique_cancers:
    subset = test_results[test_results["cancer_type"] == ct].copy()
    optimal_threshold, best_p_val = find_optimal_cutpoint(subset, risk_column="risk",
                                                          time_column="OS.time",
                                                          event_column="OS")
    if optimal_threshold is None:
        print(f"No optimal cutpoint found for {ct}; using median instead.")
        optimal_threshold = subset["risk"].median()
    else:
        print(f"Optimal threshold for {ct}: {optimal_threshold:.3f} (log-rank p-value = {best_p_val:.4f})")
    subset["risk_group"] = subset["risk"].apply(lambda x: "high" if x >= optimal_threshold else "low")
    
    kmf_low = KaplanMeierFitter()
    kmf_high = KaplanMeierFitter()
    
    low_group = subset[subset["risk_group"] == "low"]
    high_group = subset[subset["risk_group"] == "high"]
    
    kmf_low.fit(low_group["OS.time"], event_observed=low_group["OS"], label=f"{ct} - Low Risk")
    kmf_high.fit(high_group["OS.time"], event_observed=high_group["OS"], label=f"{ct} - High Risk")
    
    ax = kmf_low.plot(ci_show=False, lw=2)
    kmf_high.plot(ax=ax, ci_show=False, lw=2)
    
    results = logrank_test(low_group["OS.time"], high_group["OS.time"],
                           event_observed_A=low_group["OS"], event_observed_B=high_group["OS"])
    print(f"Log-rank test p-value for {ct}: {results.p_value}")

plt.title("Combined Kaplan-Meier Curves for Cancer Types (Data-Driven Risk Stratification)")
plt.xlabel("Time (days)")
plt.ylabel("Survival Probability")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sksurv.ensemble import RandomSurvivalForest
from sksurv.metrics import cumulative_dynamic_auc
from sklearn.model_selection import train_test_split, KFold, learning_curve
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test
import optuna
import time

# -------------------------------
# Function to Plot Learning Curve
# -------------------------------
def plot_learning_curve(model, X, y, cv=5):
    train_sizes, train_scores, test_scores = learning_curve(
        model, X, y, cv=cv, n_jobs=-1, train_sizes=np.linspace(0.1, 1.0, 5))
    
    train_mean = np.mean(train_scores, axis=1)
    train_std = np.std(train_scores, axis=1)
    test_mean = np.mean(test_scores, axis=1)
    test_std = np.std(test_scores, axis=1)
    
    plt.figure(figsize=(8, 6))
    plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.1, color="r")
    plt.fill_between(train_sizes, test_mean - test_std, test_mean + test_std, alpha=0.1, color="g")
    plt.plot(train_sizes, train_mean, 'o-', color="r", label="Training score")
    plt.plot(train_sizes, test_mean, 'o-', color="g", label="Cross-validation score")
    
    plt.xlabel("Training examples")
    plt.ylabel("Score")
    plt.legend(loc="best")
    plt.title("Learning Curve")
    plt.show()

# -------------------------------
# 1. Load the Merged Survival Data
# -------------------------------
data_path = "/Users/nijat/Desktop/Final/merged_survival_analysis.csv"
df = pd.read_csv(data_path)
print("Merged survival analysis data shape:", df.shape)

# -------------------------------
# 2. Define Features and Outcome
# -------------------------------

# (a) Methylation features: all columns starting with "cg"
methylation_cols = [col for col in df.columns if col.startswith("cg")]

# (b) Gene data columns
gene_cols = ["MCL1", "MYCN", "GLI2", "EGFR", "CDK6", "MET", "FGFR1", "MYC", 
             "CDKN2A.B", "PTCH1", "PTEN", "CCND1", "KRAS", "CDK4", "MDM2", 
             "RB1", "TP53", "NF1", "ERBB2", "CCNE1"]

# (c) MGMT column 
feature_cols = methylation_cols + gene_cols + ["MGMT"]

# (d) Survival outcome columns
surv_cols = ["OS", "OS.time", "classification.2021_simplified.labels"]

df = df.dropna(subset=feature_cols + surv_cols)
print("Data shape after dropping missing values:", df.shape)

# Create a structured array for the survival outcome
y = np.array([(bool(row["OS"]), row["OS.time"]) for _, row in df.iterrows()], dtype=[("event", "bool"), ("time", "float")])
X = df[feature_cols].copy()
cancer_type = df["classification.2021_simplified.labels"].astype(str)

# -------------------------------
# 3. Train-Test Split (Stratified)
# -------------------------------
stratify_label = cancer_type + "_" + df["OS"].astype(str)
X_train, X_test, y_train, y_test, cancer_train, cancer_test = train_test_split(
    X, y, cancer_type, test_size=0.3, random_state=42, stratify=stratify_label)
print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)

# -------------------------------
# 4. Hyperparameter Tuning with Optuna
# -------------------------------
def objective(trial):
    n_estimators = trial.suggest_int("n_estimators", 50, 200)
    min_samples_split = trial.suggest_int("min_samples_split", 2, 20)
    min_samples_leaf = trial.suggest_int("min_samples_leaf", 1, 20)
    max_features = trial.suggest_categorical("max_features", ["sqrt", "log2", None])
    
    rsf = RandomSurvivalForest(
        n_estimators=n_estimators,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        max_features=max_features,
        n_jobs=-1,
        random_state=42
    )
    
    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    c_indexes = []
    for train_idx, val_idx in cv.split(X_train):
        X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
        y_tr, y_val = y_train[train_idx], y_train[val_idx]
        rsf.fit(X_tr, y_tr)
        c_index = rsf.score(X_val, y_val)
        c_indexes.append(c_index)
    return np.mean(c_indexes)

print("Starting hyperparameter tuning with Optuna...")
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=100, show_progress_bar=True)
print("Best hyperparameters:", study.best_trial.params)

# -------------------------------
# 5. Train Final RSF Model with Best Parameters
# -------------------------------
best_params = study.best_trial.params
rsf_final = RandomSurvivalForest(
    n_estimators=best_params["n_estimators"],
    min_samples_split=best_params["min_samples_split"],
    min_samples_leaf=best_params["min_samples_leaf"],
    max_features=best_params["max_features"],
    n_jobs=-1,
    random_state=42
)
start_model_time = time.time()
rsf_final.fit(X_train, y_train)
print("Final Random Survival Forest model trained in {:.2f} seconds.".format(time.time()-start_model_time))

# Plot Learning Curve
plot_learning_curve(rsf_final, X_train, y_train, cv=5)

# -------------------------------
# 6. Model Evaluation
# -------------------------------
c_index_final = rsf_final.score(X_test, y_test)
print("Overall C-index on test set:", c_index_final)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import shap

# Set Seaborn style
sns.set_style("whitegrid")

# Example gene list (adjust as needed)
gene_cols = ["MCL1", "MYCN", "GLI2", "EGFR", "CDK6", "MET", "FGFR1", "MYC", 
             "CDKN2A.B", "PTCH1", "PTEN", "CCND1", "KRAS", "CDK4", "MDM2", 
             "RB1", "TP53", "NF1", "ERBB2", "CCNE1"]

def assign_group(feature):
    """Assign a feature to a group based on its name."""
    if feature.startswith("cg"):
        return "Methylation"
    elif feature in gene_cols:
        return "Gene"
    elif feature == "MGMT":
        return "MGMT"
    else:
        return "Other"

# Define optimal thresholds for risk stratification
optimal_thresholds = {
    "astrocytoma": 47.654,
    "glioblastoma": 145.009,
    "oligodendroglioma": 27.289
}

def classify_new_patient(new_data, cancer_type, rsf_model, thresholds, explainer):
    """
    Classify a new patient given their data.
    """
    risk_score = rsf_model.predict(new_data)[0]
    print(f"Risk score for new patient: {risk_score:.3f}")
    
    threshold = thresholds.get(cancer_type, None)
    risk_group = "high" if threshold and risk_score >= threshold else "low"
    print(f"Assigned risk group for {cancer_type}: {risk_group} (Threshold: {threshold})")
    
    new_shap_values = explainer.shap_values(new_data)
    new_shap_df = pd.DataFrame(new_shap_values, columns=new_data.columns)
    
    new_shap_df_T = new_shap_df.T
    new_shap_df_T.columns = ["shap_value"]
    new_shap_df_T["group"] = new_shap_df_T.index.map(assign_group)
    
    agg_new = new_shap_df_T.groupby("group")["shap_value"].agg(lambda x: np.sum(np.abs(x))).reset_index()
    total_abs = agg_new["shap_value"].sum()
    agg_new["relative_contribution"] = 100 * agg_new["shap_value"] / total_abs if total_abs > 0 else 0
    
    print("\nRelative contribution by feature group for the new patient:")
    print(agg_new)
    
    plt.figure(figsize=(8, 5))
    sns.barplot(x="group", y="relative_contribution", hue="group", data=agg_new, palette="coolwarm", legend=False)
    plt.xlabel("Feature Group")
    plt.ylabel("Relative Contribution (%)")
    plt.title("Feature Contribution to Risk Score")
    plt.show()
    
    if total_abs > 0:
        shap.summary_plot(new_shap_values, new_data, plot_type="beeswarm")
    
    return risk_score, risk_group, agg_new

X_sample = X_train.sample(n=100, random_state=42)
explainer = shap.KernelExplainer(rsf_final.predict, X_sample)

new_patient_data = X_test.sample(n=1)
new_patient_cancer = cancer_test.loc[new_patient_data.index[0]]

print(f"\n--- Classifying New Patient (Cancer Type: {new_patient_cancer}) ---")
risk, risk_group, group_importance = classify_new_patient(new_patient_data, new_patient_cancer, rsf_final, optimal_thresholds, explainer)
```
